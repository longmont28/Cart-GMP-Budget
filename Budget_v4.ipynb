# GMP CAR-T Budget v4 — GitHub/Colab compatible
Clones the GitHub repository and runs the single-construct, single-planned-lot scenario model. Prices remain unquoted until verified.


```python
import os, subprocess, sys
repo = "https://github.com/longmont28/Cart-GMP-Budget.git"
folder = "/content/Cart-GMP-Budget"
if not os.path.isdir(folder + "/.git"):
    subprocess.run(["git", "clone", repo, folder], check=True)
else:
    subprocess.run(["git", "-C", folder, "pull", "--ff-only"], check=True)
sys.path.insert(0, folder)
from model import run
for case in ("low", "base", "high"):
    x = run(case)
    print(case, "screened:", x["planned_screened"], "assurance:", round(x["probability_at_least_target"], 3), "vector lots:", x["capacity"]["vector_attempts"], "vector gap:", x["capacity"]["vector_demand_gap"], "approved budget:", x["approved_total_sek"])
```
